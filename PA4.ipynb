Task 1: Input Positive Integer and Factors (14 pts)
A. Let a user input a positive integer.  You are to then print all positive integers that are a factor of the input number.  Sample input/output follows:
Please input a positive integer
: 12
1 is a factor of 12
2 is a factor of 12
3 is a factor of 12
4 is a factor of 12
6 is a factor of 12
12 is a factor of 12

```python
number = int(input("Enter a positive integer: "))

for i in range(1, number + 1):
    if number % i == 0:
        print(f"{i} is a factor of {number}")
```

B. Let a user input a positive  integer.  You are to then determine whether or not the number is prime and print a corresponding response to the user (e.g. “Your number is prime” or “Your number is not prime”. )
Hint: a relatively easy solution to this problem uses an approach closely related to that usable for Part A.

```python
number = int(input("Enter a positive integer: "))
factors = 0

for i in range(1, number + 1):
    if number % i == 0:
        factors = factors + 1

if factors == 2:
    print("Your number is prime")
else:
    print("Your number is not prime")
```

Task 2: Language Translator (10 pts)
Write a program that implements a translation dictionary using a Python dictionary that maps 10 English words to 10 words in another language (e.g. Spanish, German, Farsi, Arabic, Japanese, etc.) of your choosing. The keys in your dictionary should be in English (all lower case), and your values should be in your target language (also all lowercase). Your program should ask for input from a user in English to translate to your target language until the user enters the empty string (no input). If a word entered by the user is not in your translation dictionary, you should let the user know and continue reading input words.  Sample input/output follows:

```python
mydic = {
    "hello": "hola",
    "goodbye": "adios",
    "yes": "si",
    "no": "no",
    "please": "por favor",
    "thank you": "gracias",
    "water": "agua",
    "green": "verde",
    "house": "casa",
    "library": "bibliotecha"
}
```

```python
word = input("Enter an English word: ").lower()

if word in mydic:
    print(mydic[word])
else:
    print("Word not found")
```

```python
word = input("Enter an English word: ").lower()

while word != "":
    if word in mydic:
        print(mydic[word])
    else:
        print("Word not found")

    word = input("Enter an English word: ").lower()
```

Task 3: Temperature Summaries (10 pts)
	Using a weather app, google search, or similar find the average high and low temperature for each month in Louisville in Farenheit.  (You may use the most recent year, overall averages, etc.) Store the temperature data as a 2-element list (value) for each month (key) in a dictionary.  
For example, the dictionary may begin as follows for the first two months of the year:
{‘January’,[43,  27], ‘February’:[49, 30]…
Allow the user to input months by name, and return the high and low temperatures for the given month, if it is an actual month.  If the user enters “quit”, the program should stop.  Sample input/output follows:
	Enter a month: I will give you the average high and low temperatures in Louisville
: January
The average high temperature in Louisville for January is 43 and the average low is 27.
Enter a month: I will give you the average high and low temperatures in Louisville
: Augus
Not a valid month
Enter a month: I will give you the average high and low temperatures in Louisville
: February
The average high temperature in Louisville for February is 49 and the average low is 30.
Enter a month: I will give you the average high and low temperatures in Louisville
: quit


```python
temps = {
    "january": [43, 27],
    "february": [49, 30],
    "march": [59, 38],
    "april": [70, 47],
    "may": [78, 57],
    "june": [86, 66],
    "july": [89, 70],
    "august": [88, 68],
    "september": [82, 61],
    "october": [71, 49],
    "november": [58, 39],
    "december": [47, 32]
}
month = input("Enter a month or quit: ").lower()

while month != "quit":
    if month in temps:
        print("Average high:", temps[month][0])
        print("Average low:", temps[month][1])
    else:
        print("Invalid month")

    month = input("Enter a month or quit: ").lower()
```

	inventory = {‘hammer’:[15, 12.49], ‘wrench’:[20, 7.99], ‘screwdriver’: [40, 3.99], ‘nailgun’:[10, 36.59], ‘padlock’:[25, 5.99], ‘roll_of_duct_tape’: [30, 4.95]} 
	Ex: In the dictionary provided, there are 20 wrenches available at 7.99 each.
	After storing the dictionary above in a variable, you are to create a menu to permit the user to select one of several options by number as follows:
1. Let the user enter an item by name – if the item is in the dictionary, return the number available and the corresponding price.  If it is not, let them know it isn’t available.
2. Let the user input a maximum budget amount – display all items with an individual cost below this budget.
3. Tell the user the total quantity of items available across all item names (here you can simply sum the total number of quantities for each entry – ex: 15 + 20 + 40 + …).
4. Tell the user the total price of all items across the inventory (here you will need to add the total costs across all items – ex: 15 x 12.49 + 20 x 7.99, etc.)
	Note: you may put your menu within a loop, but do not have to do so for this problem.


```python
inventory = {
    "hammer": [15, 12.49],
    "wrench": [20, 7.99],
    "screwdriver": [40, 3.99],
    "nailgun": [10, 36.59],
    "padlock": [25, 5.99],
    "roll_of_duct_tape": [30, 4.95]
}
print("Inventory Menu")
print("1. Look up an item")
print("2. Find items under a certain price")
print("3. Total number of items")
print("4. Total inventory value")

choice = int(input("Enter your choice (1-4): "))
if choice == 1:
    item = input("Enter an item: ").lower()

    if item in inventory:
        print("Quantity:", inventory[item][0])
        print("Price: $", inventory[item][1])
    else:
        print("Item not found")
elif choice == 2:
    price = float(input("Enter your maximum price: "))

    for item in inventory:
        if inventory[item][1] < price:
            print(item, "$", inventory[item][1])
elif choice == 3:
    total = 0

    for item in inventory:
        total = total + inventory[item][0]

    print("Total number of items:", total)
elif choice == 4:
    total = 0

    for item in inventory:
        total = total + inventory[item][0] * inventory[item][1]

print(f"Total inventory value: ${total:.2f}")
```
