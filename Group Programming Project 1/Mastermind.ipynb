```python

import random

record = {"wins": 0, "losses": 0}
play_again = "y"

while play_again == "y":

   
    secret = ""
    for i in range(4):
        secret = secret + str(random.randint(0, 9))

    print("I picked a secret 4-digit number. You have 10 guesses.")
    won=False
    

   
    for turn in range(1, 11):
        guess = input("Guess " + str(turn) + ": ")
        while len(guess) != 4:
            guess = input("Please enter exactly 4 digits: ")

        
        right_spot = 0
        secret_left = ""
        guess_left = ""
        for i in range(4):
            if guess[i] == secret[i]:
                right_spot= right_spot +1
            else:
                secret_left = secret_left + secret[i]
                guess_left = guess_left + guess[i]

                
        wrong_spot = 0
        for digit in range(0,10):
            in_secret = secret_left.count(str(digit))
            in_guess = guess_left.count(str(digit))
            if in_secret < in_guess:
                wrong_spot = wrong_spot + in_secret
            else:
                wrong_spot = wrong_spot + in_guess

        
        if right_spot == 4:
            print("You win! The number was", secret)
            won = True
            break

        print(right_spot, "correct digit(s) in the correct position")
        print(wrong_spot, "correct digit(s) in the wrong position")

    
    if won:
        record["wins"] = record["wins"] + 1
    else:
        print("You lose! The number was", secret)
        record["losses"] = record["losses"] + 1

    print("Record: Wins =", record["wins"], " Losses =", record["losses"])

    #
    play_again = input("Play again? (y/n): ")

print("Thanks for playing!")
```
