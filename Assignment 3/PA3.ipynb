## Task 1: Types of Triangles

Take user input representing three sides of a triangle. Determine whether the triangle is equilateral, isosceles, or scalene.

```python
side1 = float(input("Enter the first side: "))
side2 = float(input("Enter the second side: "))
side3 = float(input("Enter the third side: "))

if side1 == side2 and side2 == side3:
    print("The triangle is equilateral.")
elif side1 == side2 or side1 == side3 or side2 == side3:
    print("The triangle is isosceles.")
else:
    print("The triangle is scalene.")
```

## Task 2: Student Grades and At-Risk Status

Read five student grade averages into a list. Calculate the class average, minimum grade, maximum grade, and determine how many students are at risk.

```python
student_grades = []

grade1 = float(input("Enter the first student's grade: "))
student_grades.append(grade1)

grade2 = float(input("Enter the second student's grade: "))
student_grades.append(grade2)

grade3 = float(input("Enter the third student's grade: "))
student_grades.append(grade3)

grade4 = float(input("Enter the fourth student's grade: "))
student_grades.append(grade4)

grade5 = float(input("Enter the fifth student's grade: "))
student_grades.append(grade5)

average = sum(student_grades) / len(student_grades)

print("Class average:", average)
print("Minimum grade:", min(student_grades))
print("Maximum grade:", max(student_grades))

at_risk = 0
number = 0

while number < len(student_grades):
    if student_grades[number] < 65:
        at_risk = at_risk + 1
    number = number + 1

print(at_risk, "students are at risk in the course.")
```

## Task 3: Ranges and Multiples

Create lists containing multiples of 2, multiples of 3, multiples of both, and multiples of either between 1 and 100.

```python
mult_of_2 = []
number = 1

while number <= 100:
    if number % 2 == 0:
        mult_of_2.append(number)
    number = number + 1

print(mult_of_2)
```

```python
mult_of_3 = []
number = 1

while number <= 100:
    if number % 3 == 0:
        mult_of_3.append(number)
    number = number + 1

print(mult_of_3)
```

```python
mult_of_both = []
number = 1

while number <= 100:
    if number % 6 == 0:
        mult_of_both.append(number)
    number = number + 1

print(mult_of_both)
```

```python
mult_of_either = []
number = 1

while number <= 100:
    if number in mult_of_2 or number in mult_of_3:
        mult_of_either.append(number)
    number = number + 1

print(mult_of_either)
```

## Task 4: Shopping List

Create a shopping list that allows the user to enter up to 7 items, then display information about the completed list.

```python
shopping_list = []
count = 0

while count < 7:
    item = input("Enter an item or type done to finish: ")

    if item == "done":
        break

    shopping_list.append(item)
    count = count + 1

print(shopping_list)
```

```python
print("Total number of items:", len(shopping_list))
```

```python
reverse_list = shopping_list.copy()
reverse_list.reverse()

print(reverse_list)
```

```python
alphabetical_list = sorted(shopping_list)

print(alphabetical_list)
```

```python
duplicate = False
count = 0

while count < len(shopping_list):
    if shopping_list.count(shopping_list[count]) >= 2:
        duplicate = True
    count = count + 1

if duplicate == True:
    print("At least one item appears more than once in the shopping list.")
else:
    print("No items appear more than once in the shopping list.")
```
