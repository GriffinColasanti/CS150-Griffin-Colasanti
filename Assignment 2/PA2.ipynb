## Task 2: Distance Conversion

Create a variable for a distance in miles, convert the distance to kilometers using the given formula, and print both distances.

```python
miles_distance = 10
kilometers = miles_distance * 1.60934

print("The distance is " + str(miles_distance) + " miles, which is the equivalent of " + str(kilometers) + " kilometers.")
```

## Task 3: Working with Running Times

Create two variables representing running times in minutes. Calculate the total time, the difference between the runs, twice the second run, and the average of both runs.

```python
run1 = 30
run2 = 27

total_time = run1 + run2
improvement = run1 - run2
twice_run2 = run2 * 2
average_time = (run1 + run2) / 2

print("Total time:", total_time)
print("Improvement:", improvement)
print("Run 2 twice:", twice_run2)
print("Average time:", average_time)
```

## Task 4: String Manipulation

## Task 5: User Input and Operations

Prompt the user to enter the lengths of the two legs of a right triangle. Use the Pythagorean theorem to calculate the hypotenuse and print the result rounded to the nearest one-hundredth.

```python
a = float(input("Enter the length of the first leg: "))
b = float(input("Enter the length of the second leg: "))

c = (a**2 + b**2)**0.5

print("The hypotenuse of your triangle is " + str(round(c, 2)) + " units.")
```
